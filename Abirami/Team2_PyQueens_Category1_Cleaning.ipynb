<h2> Detect the outliers for all the major parameters in each of the files. </h2>

Detect the outliers for all the five patients in each file. For that we give upper limits and lower limits for the parameters which we need to handle and only those files which has outliers will be displayed. For glucose we give the upper limit as 500 and lower limit as 40. For the heart rate, upper limit is 220 and lower limit is 30. Similarly for the other parameters also we fix the upper and lower limits. Any patient whose values show either above or below this range, there might be a sensor error.

```python
def identify_real_outliers(df):
    df['outlier'] = False
    df['outlier_column'] = ""

    for idx, row in df.iterrows():

        # Glucose
        if row['Glucose'] < 40 or row['Glucose'] >500:
            df.at[idx, 'outlier'] = True
            df.at[idx, 'outlier_column'] += "Glucose, "

        # Heart rate
        if row['Heart_rate'] < 30 or row['Heart_rate'] > 220:
            df.at[idx, 'outlier'] = True
            df.at[idx, 'outlier_column'] += "Heart_rate, "

        # Steps
        if row['Steps'] < 0 or row['Steps'] > 6000:
            df.at[idx, 'outlier'] = True
            df.at[idx, 'outlier_column'] += "Steps, "

        # Bolus insulin
        if row['Bolus_volume_delivered'] < 0 or row['Bolus_volume_delivered'] > 25:
            df.at[idx, 'outlier'] = True
            df.at[idx, 'outlier_column'] += "Bolus_volume_delivered, "

        # Carb input
        if row['Carb_input'] < 0 or row['Carb_input'] > 200:
            df.at[idx, 'outlier'] = True
            df.at[idx, 'outlier_column'] += "Carb_input, "

    # Print message
    if not df['outlier'].any():
        print("No outliers in the specified columns")
    else:
        print("Outliers found in the specified columns")

    # Return only the outlier rows
    return df[df['outlier'] == True]
```

<h2 style="color: purple;"> PATIENT_ID -HUPA0019P,HUPA0020P,HUPA0021P,HUPA0022P,HUPA0023P IN TO A SINGLE FILE FOR CLEANING </h2>

```python
import pandas as pd
import glob
import os

# Get 5 CSV files from the folder
cgm_5files_hg = glob.glob("C:/Users/Owner/Downloads/Data_5file/*.csv")

# Add Patient_ID column & merge all 5 files
dfs = []
for file in cgm_5files_hg:
    if os.path.isfile(file):
        df = pd.read_csv(file, sep=';')
        df['patient_id'] = os.path.basename(file).replace('.csv', '')
        dfs.append(df)   # ← this must be inside the IF block

# Merge all files
df_CGM5_hg = pd.concat(dfs, ignore_index=True)

# Move patient_id to first column
cols = ['patient_id'] + [col for col in df_CGM5_hg.columns if col != 'patient_id']
df_CGM5_hg = df_CGM5_hg[cols]

df_CGM5_hg
```

### To improve readability and maintain consistent formatting across the dataset, all column names were converted to Title Case. This ensures that each word in the column name starts with a capital letter, making the dataset easier to understand and more visualization.

```python
# Convert all column names to title case
df_CGM5_hg.columns = df_CGM5_hg.columns.str.title()

# Check Existing datatype of columns
df_CGM5_hg.dtypes

# Convert 'time' column from object to datetime format for time-based operation
df_CGM5_hg['Time'] = pd.to_datetime(df_CGM5_hg['Time'])

# Convert 'steps' column from float64 to int as steps are always whole numbers
df_CGM5_hg['Steps'] = df_CGM5_hg['Steps'].astype(int)

df_CGM5_hg.dtypes
```

### To ensure consistency and improve readability in the dataset, all columns with the data type float64 were rounded to 2 decimal places. This is especially useful when working with measurements such as glucose levels, BMI, insulin values, or any continuous numeric data.The code identifies all float64 columns using `select_dtypes()` and then applies the `round(2)` function to those columns only.

```python
# Round all float64 columns to 2 decimal places
float_cols = df_CGM5_hg.select_dtypes(include='float64').columns
df_CGM5_hg[float_cols] = df_CGM5_hg[float_cols].round(2)

df_CGM5_hg.tail()
```

### Duplicate rows can cause incorrect analysis, biased statistics, and misleading results.To ensure data quality, we check for duplicate entries in the dataset.

```python
#Duplicate value check
print("Total duplicates:", df_CGM5_hg.duplicated().sum())
```

### Null values (missing data) can affect analysis, calculations, and model performance.To ensure data quality, we checked for missing values in each column of the dataset.

```python
#Null Value
df_CGM5_hg.isnull().sum()
```

### Negative values are not valid for this dataset because variables such as glucose level, insulin dose, steps, BMI, and heart rate cannot logically be negative.To ensure data quality, we checked all columns for negative values and handled them appropriately.

```python
# No column should have negative values
numeric_cols = df_CGM5_hg.select_dtypes(include=['float64', 'int64']).columns
for col in numeric_cols:
    neg = (df_CGM5_hg[col] < 0).sum()
    print(f"{col}: {neg} negative values")
```

### Whitespace in a dataset can cause issues such as incorrect grouping, mismatched categories, failed merges, and inaccurate analysis.To ensure data consistency, we checked all columns for the presence of whitespace characters (spaces, tabs, newlines).

```python
# Check whitespace
print("Whitespace check in all columns:")
for col in df_CGM5_hg.columns:
    if df_CGM5_hg[col].dtype == object:  # only text columns
        count = df_CGM5_hg[col].astype(str).str.contains(r'^\s|\s$').sum()
        print(f"  {col}: {count} rows with whitespace")
```

### To perform accurate time‑series analysis, the dataset must be ordered chronologically for each patient.Sorting ensures that glucose readings, insulin doses, steps, and other measurements follow the correct timeline.

```python
# Sort data by patient and time for proper time series analysis
df_CGM5_hg = df_CGM5_hg.sort_values(['Patient_Id', 'Time']).reset_index(drop=True)
df_CGM5_hg.head()
```
