Detect the outliers for all the five patients in each file. For that we give upper limits and lower limits for the parameters which we need to handle and only those files which has outliers will be displayed.

```python
def identify_real_outliers(df):
    df['outlier'] = False
    df['outlier_column'] = ""

    for idx, row in df.iterrows():

        # Glucose
        if row['Glucose'] < 40 or row['Glucose'] >500:
            df.at[idx, 'outlier'] = True
            df.at[idx, 'outlier_column'] += "Glucose, "

        # Heart rate
        if row['Heart_rate'] < 30 or row['Heart_rate'] > 220:
            df.at[idx, 'outlier'] = True
            df.at[idx, 'outlier_column'] += "Heart_rate, "

        # Steps
        if row['Steps'] < 0 or row['Steps'] > 6000:
            df.at[idx, 'outlier'] = True
            df.at[idx, 'outlier_column'] += "Steps, "

        # Bolus insulin
        if row['Bolus_volume_delivered'] < 0 or row['Bolus_volume_delivered'] > 25:
            df.at[idx, 'outlier'] = True
            df.at[idx, 'outlier_column'] += "Bolus_volume_delivered, "

        # Carb input
        if row['Carb_input'] < 0 or row['Carb_input'] > 200:
            df.at[idx, 'outlier'] = True
            df.at[idx, 'outlier_column'] += "Carb_input, "

    # ⭐ Print message
    if not df['outlier'].any():
        print("No outliers in the specified columns")
    else:
        print("Outliers found in the specified columns")

    # ⭐ Return only the outlier rows
    return df[df['outlier'] == True]
```

```python
all_dfs = {
    "HUPA11": df_HUPA11,
    "HUPA14": df_HUPA14,
    "HUPA15": df_HUPA15,
    "HUPA16": df_HUPA16,
    "HUPA17": df_HUPA17
}
```

```python
outlier_results = {}

for name, df in all_dfs.items():
    print(f"\nChecking outliers for {name}...")
    outlier_results[name] = identify_real_outliers(df)
```
