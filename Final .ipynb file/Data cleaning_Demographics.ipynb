```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\vidya\OneDrive\Desktop\DATA ANALYST\NUMPY NINJA\GITHUB\Team2_-PyQueens_-Python-Hackathon-_MAY-2026\Final cleaned csv files\Final cleaned csv files\Team2_PyQueens_Merged_Final.csv"
)

print("Shape:", df.shape)
print("\nData Types:")
print(df.dtypes)
print("\nFirst 3 rows:")
df.head(3)
```

```python
# ── Fix Data Types ────────────────────────────────────────────────────────────
df['Time']     = pd.to_datetime(df['Time'], format='mixed')
df['Gender']   = df['Gender'].astype('category')
df['Race']     = df['Race'].astype('category')
df['Patient_Id'] = df['Patient_Id'].astype('string')

# Numeric columns — ensure float64
float_cols = [
    'Glucose', 'Calories', 'Heart_Rate', 'Steps',
    'Basal_Rate', 'Bolus_Volume_Delivered', 'Carb_Input',
    'Average Sleep Duration (hrs)', 'Sleep Quality (1-10)'
]
df[float_cols] = df[float_cols].astype(float)

# Integer columns
df['Age']                       = df['Age'].astype(int)
df['% with Sleep Disturbances'] = df['% with Sleep Disturbances'].astype(int)

# ── Verify ────────────────────────────────────────────────────────────────────
print("Updated Data Types:")
print(df.dtypes)
```

```python
import pandas as pd

# ── Step 1: Load both files ──
df_main = pd.read_csv(r"C:\Users\vidya\OneDrive\Desktop\DATA ANALYST\NUMPY NINJA\GITHUB\Team2_-PyQueens_-Python-Hackathon-_MAY-2026\Final cleaned csv files\Final cleaned csv files\Team2_PyQueens_Final_Cleaned_File.csv")

df_demo = pd.read_csv(r"C:\Users\vidya\OneDrive\Desktop\DATA ANALYST\NUMPY NINJA\GITHUB\Team2_-PyQueens_-Python-Hackathon-_MAY-2026\Final cleaned csv files\Final cleaned csv files\T1DM_patient_sleep_demographics_with_race.csv")

print("File 1 shape :", df_main.shape)
print("File 1 columns:", list(df_main.columns))
print("\nFile 2 shape :", df_demo.shape)
print("File 2 columns:", list(df_demo.columns))

# ── Step 2: Standardize Patient ID column name ──
df_demo.rename(columns={'Patient_ID': 'Patient_Id'}, inplace=True)

# ── Step 3: Merge on Patient_Id (many-to-one) ──
# df_main has 309392 rows (sensor readings)
# df_demo has 25 rows (one per patient)
# Each sensor row gets the demographic info of that patient
merged_df = pd.merge(df_main, df_demo, on='Patient_Id', how='left')

# ── Step 4: Validate merge ──
print(f"\n✅ Merged DataFrame:")
print(f"   Shape          : {merged_df.shape[0]} rows × {merged_df.shape[1]} columns")
print(f"   Total Patients : {merged_df['Patient_Id'].nunique()}")
print(f"   Columns        : {list(merged_df.columns)}")
print(f"   Missing after merge: {merged_df.isnull().sum().sum()}")
print(f"\nSample:")
print(merged_df[['Patient_Id','Time','Glucose','Steps','Age','Gender',
                  'Average Sleep Duration (hrs)','Sleep Quality (1-10)']].head(5))

# ── Step 5: Save merged file ──
output_path = r"C:\Users\vidya\OneDrive\Desktop\DATA ANALYST\NUMPY NINJA\GITHUB\Team2_-PyQueens_-Python-Hackathon-_MAY-2026\Final cleaned csv files\Final cleaned csv files\Team2_PyQueens_Merged_Final.csv"

merged_df.to_csv(output_path, index=False)
print(f"\n✅ Merged file saved as: Team2_PyQueens_Merged_Final.csv")
print(f"   Rows    : {len(merged_df)}")
print(f"   Columns : {merged_df.shape[1]}")
```
