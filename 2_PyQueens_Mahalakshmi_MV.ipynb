```python
import pandas as pd
import glob
import os

cgm_5files_mv = glob.glob("C:/python_hackathon/HUPA-UC Diabetes Dataset-20250820T010637Z-1-001/HUPA-UC Diabetes Dataset/CGM5_MV/*.csv")

dfs = []
for file in cgm_5files_mv :
    if os.path.isfile(file):
        df = pd.read_csv(file, sep=';')
        df['patient_id'] = os.path.basename(file).replace('.csv', '')
        dfs.append(df)


df_CGM5_MV= pd.concat(dfs, ignore_index=True)

# Move patient_id to first column
cols = ['patient_id'] + [col for col in df_CGM5_MV.columns if col != 'patient_id']
df_CGM5_MV = df_CGM5_MV[cols]

df_CGM5_MV
```

```python
df_CGM5_MV[df_CGM5_MV.duplicated(subset="patient_id", keep=False)]
df_CGM5_MV[df_CGM5_MV.duplicated(subset="time", keep=False)]
df_CGM5_MV[df_CGM5_MV.duplicated(subset="glucose", keep=False)]
df_CGM5_MV[df_CGM5_MV.duplicated(subset="heart_rate", keep=False)]
df_CGM5_MV[df_CGM5_MV.duplicated(subset="steps", keep=False)]
df_CGM5_MV[df_CGM5_MV.duplicated(subset="basal_rate", keep=False)]
df_CGM5_MV[df_CGM5_MV.duplicated(subset="bolus_volume_delivered", keep=False)]
df_CGM5_MV[df_CGM5_MV.duplicated(subset="carb_input", keep=False)]
```

```python
df_CGM5_MV.columns = df_CGM5_MV.columns.str.title()
```

```python
df_CGM5_MV = df_CGM5_MV.round(2)
```

```python
df_CGM5_MV['Glucose_Outlier'] = (df_CGM5_MV['Glucose'] < 40) | (df_CGM5_MV['Glucose'] > 400)
```
